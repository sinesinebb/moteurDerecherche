```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Paramètres BM25
k1 = 1.5
b = 0.75

# Fréquence du terme dans le document
tf_values = np.arange(1, 51)

# Longueur du document et longueur moyenne (on fixe |d|/avgdl = 1 pour simplifier)
doc_length_ratio = 1.0

# Calcul du score BM25 simplifié (IDF constant)
# IDF fictif pour l'illustration
idf = 1.0
bm25 = idf * (tf_values * (k1 + 1)) / (tf_values + k1 * (1 - b + b * doc_length_ratio))

# TF‑IDF (proportionnel à la fréquence)
tfidf = idf * tf_values
```

```python
# Tracé
plt.figure(figsize=(8, 5))
plt.plot(tf_values, tfidf, 'b-', linewidth=2, label='TF‑IDF (linéaire)')
plt.plot(tf_values, bm25, 'r--', linewidth=2, label='BM25 (saturation)')
plt.xlabel('Fréquence du terme dans le document', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.title('Comparaison TF‑IDF vs BM25\n(avec $k_1 = 1.5$, $b = 0.75$)', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('tfidf_vs_bm25.png', dpi=150, bbox_inches='tight')
plt.show()
```
